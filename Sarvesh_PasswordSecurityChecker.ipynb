## 1. Import required libraries
This project uses only Python's standard library, so no extra packages are required.

```python
import re
```

## 2. Define the password checker
The checker gives a simple strength rating. This is an educational estimate, not a guarantee that a password is secure.

```python
COMMON_PASSWORDS = {
    "password", "password123", "123456", "12345678",
    "qwerty", "admin", "letmein", "iloveyou"
}

def check_password(password):
    """Return a basic strength assessment for an example password."""
    score = 0
    feedback = []

    if len(password) >= 12:
        score += 2
    elif len(password) >= 8:
        score += 1
    else:
        feedback.append("Use at least 12 characters where possible.")

    if re.search(r"[a-z]", password):
        score += 1
    else:
        feedback.append("Add lowercase letters.")

    if re.search(r"[A-Z]", password):
        score += 1
    else:
        feedback.append("Add uppercase letters.")

    if re.search(r"\d", password):
        score += 1
    else:
        feedback.append("Consider including numbers.")

    if re.search(r"[^A-Za-z0-9]", password):
        score += 1
    else:
        feedback.append("Consider including symbols.")

    if password.lower() in COMMON_PASSWORDS:
        score = 0
        feedback.insert(0, "This is a commonly used password. Do not use it.")

    if score <= 2:
        strength = "Weak"
    elif score <= 4:
        strength = "Moderate"
    else:
        strength = "Strong"

    if not feedback:
        feedback.append("Good mix of length and character types. Avoid reusing passwords.")

    return {"strength": strength, "score": score, "feedback": feedback}
```

## 3. Test with sample passwords
These are examples only. Never use a real password in a shared notebook.

```python
sample_passwords = ["123456", "hello123", "BlueRiver!47Cloud"]
for sample in sample_passwords:
    result = check_password(sample)
    print(f"Example: {sample!r}")
    print(f"Strength: {result['strength']} (score {result['score']}/6)")
    for item in result["feedback"]:
        print(" -", item)
    print()
```

## 4. Try your own example
Use a made-up test password only. The code checks it in memory and does not save it.

```python
test_password = input("Enter a SAMPLE password (not a real password): ")
result = check_password(test_password)
print("Strength:", result["strength"])
print("Score:", str(result["score"]) + "/6")
print("Suggestions:")
for item in result["feedback"]:
    print("-", item)
```

## 5. Limitations and safety
- This is a basic educational checker, not a professional password-auditing tool.
- It checks only a small built-in list of common passwords.
- A strong-looking password may still be exposed or reused elsewhere.
- Do not paste real passwords into notebooks, websites, or public repositories.
